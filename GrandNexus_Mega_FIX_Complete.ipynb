````python
import os
import sys

print("=" * 80)
print("🚀 MEGA PATCH DE RESTAURATION : RETOUR VERS LE FUTUR (V1 -> V5)")
print("=" * 80)

# Création des dossiers nécessaires
os.makedirs("grandnexus/memory", exist_ok=True)
os.makedirs("grandnexus/reasoning", exist_ok=True)

print("1️⃣ Restauration de la CURIOSITÉ (ActiveLearning V2.1)...")
code_al = '''
import logging
import random
from typing import List, Dict, Any, Optional

class ActiveLearningModule:
    def __init__(self):
        self.logger = logging.getLogger("ActiveLearning")
        self.asked_questions = set()
        self.question_templates = {
            "clarification": ["What exactly defines {concept}?", "Could you elaborate on the nature of {concept}?"],
            "relation": ["How does {concept1} interact with {concept2}?", "Is there a causal link between {concept1} and {concept2}?"],
            "mechanism": ["By what mechanism does {concept} operate?", "What are the underlying rules of {concept}?"],
            "difference": ["What distinguishes {concept1} from {concept2}?", "Are {concept1} and {concept2} fundamentally different?"]
        }

    def generate_question(self, working_memory: List[Any], graph_nodes: List[str], uncertainty: float, mode: str) -> Optional[str]:
        try:
            wm_concepts = [str(item.content) for item in working_memory if hasattr(item, 'content')]
            graph_nodes = [str(n) for n in graph_nodes]
            available_concepts = list(set(wm_concepts + graph_nodes))
            try: uncertainty = float(uncertainty)
            except: uncertainty = 0.5
            if not available_concepts: return None
            candidates = []
            for _ in range(15):
                q_type = random.choice(list(self.question_templates.keys()))
                if q_type in ["relation", "difference"] and len(available_concepts) < 2: q_type = "clarification"
                if q_type in ["relation", "difference"]:
                    concepts = random.sample(available_concepts, 2)
                    text = random.choice(self.question_templates[q_type]).format(concept1=concepts[0], concept2=concepts[1])
                else:
                    concepts = [random.choice(available_concepts)]
                    text = random.choice(self.question_templates[q_type]).format(concept=concepts[0])
                if text in self.asked_questions: continue
                score = self._calculate_eig(q_type, concepts, uncertainty, wm_concepts)
                candidates.append((text, float(score)))
            if not candidates: return None
            best_question, best_score = max(candidates, key=lambda x: x[1])
            self.asked_questions.add(best_question)
            return best_question
        except Exception as e:
            self.logger.error(f"Error: {e}")
            return None

    def _calculate_eig(self, q_type, concepts, uncertainty, wm_focus):
        score = 0.0
        for c in concepts:
            if c in wm_focus: score += 1.0
        if uncertainty > 0.8 and q_type == "clarification": score += 2.0
        elif uncertainty < 0.3 and q_type in ["relation", "difference"]: score += 2.0
        score += random.uniform(0, 0.5)
        return score
'''
with open("grandnexus/active_learning.py", "w") as f: f.write(code_al)

print("2️⃣ Restauration de la MÉMOIRE DE TRAVAIL (WorkingMemory V2.1)...")
code_wm = """
import logging
import time
from typing import List, Any, Optional

class MemoryItem:
    def __init__(self, content, memory_type="concept", activation=1.0, metadata=None):
        self.content = content
        self.memory_type = memory_type
        self.activation = float(activation)
        self.creation_time = time.time()
        self.last_access = time.time()
        self.metadata = metadata or {}
    def __repr__(self): return f"MemoryItem('{self.content}', {self.activation:.2f})"

class WorkingMemory:
    def __init__(self, capacity=10, decay_rate=0.05):
        self.items = []
        self.capacity = capacity
        self.base_decay_rate = decay_rate
        self.current_context = set()
    def add(self, content, memory_type="concept", activation=1.0, metadata=None, **kwargs):
        if 'importance' in kwargs: activation = kwargs['importance']
        self.current_context.add(str(content))
        for item in self.items:
            if item.content == content:
                item.activation = 1.0
                item.last_access = time.time()
                return
        self.items.append(MemoryItem(content, memory_type, activation, metadata))
        self._enforce_capacity()
    def get_all(self) -> List[Any]: return self.items
    def get_top_k(self, k=5, **kwargs) -> List[Any]:
        return sorted(self.items, key=lambda x: x.activation, reverse=True)[:k]
    def remove(self, content): self.items = [i for i in self.items if i.content != content]
    def _enforce_capacity(self):
        if len(self.items) > self.capacity:
            self.items.sort(key=lambda x: x.activation, reverse=True)
            self.items = self.items[:self.capacity]
    def decay(self, time_delta=None):
        factor = 1.0
        if time_delta is not None: factor = max(1.0, float(time_delta))
        for item in self.items:
            rate = self.base_decay_rate * (0.1 if item.content in self.current_context else 1.0) * factor
            item.activation -= rate
        self.items = [i for i in self.items if i.activation > 0.05]
        if len(self.current_context) > 20: self.current_context.clear()
ContextAwareWorkingMemory = WorkingMemory
"""
with open("grandnexus/memory/working.py", "w") as f: f.write(code_wm)

print("3️⃣ Restauration de la MÉMOIRE SÉMANTIQUE (SemanticMemory V2.2)...")
code_sm = '''
import logging
import time
from typing import Dict, List, Any, Tuple
from enum import Enum

class EdgeType(Enum):
    IS_A = "is_a"; HAS_A = "has_a"; PART_OF = "part_of"; CAUSES = "causes"
    CORRELATES_WITH = "correlates_with"; RELATED_TO = "related_to"

class SemanticNode:
    def __init__(self, name: str, node_type: str = "concept"):
        self.name = name; self.type = node_type; self.activation = 1.0
        self.creation_time = time.time(); self.last_access = time.time(); self.access_count = 1

class SemanticEdge:
    def __init__(self, source: str, target: str, edge_type: EdgeType, weight: float = 1.0):
        self.source = source; self.target = target; self.type = edge_type
        self.weight = float(weight); self.creation_time = time.time()

class DeepSemanticMemory:
    def __init__(self, decay_rate=0.01, **kwargs):
        self.nodes = {}; self.edges = {}; self.decay_rate = decay_rate; self.total_nodes = 0
    def add_concept(self, name: str, node_type: str = "concept", **kwargs):
        name = str(name).lower().strip()
        if not name: return
        if name in self.nodes:
            self.nodes[name].activation = 1.0
            self.nodes[name].access_count += 1
        else:
            self.nodes[name] = SemanticNode(name, node_type)
            self.edges[name] = []
            self.total_nodes += 1
    def add_relation(self, source: str, target: str, relation_type: str = "related_to", weight: float = 1.0):
        source = str(source).lower().strip(); target = str(target).lower().strip()
        if source not in self.nodes: self.add_concept(source)
        if target not in self.nodes: self.add_concept(target)
        try: edge_enum = EdgeType(relation_type)
        except: edge_enum = EdgeType.RELATED_TO
        existing = False
        for i, (tgt, edge) in enumerate(self.edges[source]):
            if tgt == target and edge.type == edge_enum:
                edge.weight = min(1.0, edge.weight + 0.1)
                existing = True; break
        if not existing:
            self.edges[source].append((target, SemanticEdge(source, target, edge_enum, weight)))
            if edge_enum in [EdgeType.RELATED_TO, EdgeType.CORRELATES_WITH]:
                self.edges[target].append((source, SemanticEdge(target, source, edge_enum, weight)))
    def get_related(self, concept: str, limit: int = 5) -> List[str]:
        concept = str(concept).lower().strip()
        if concept not in self.nodes: return []
        related = sorted(self.edges[concept], key=lambda x: x[1].weight, reverse=True)
        return [target for target, edge in related[:limit]]
    def get_status(self) -> Dict[str, Any]:
        return {"module": "DeepSemanticMemory", "total_nodes": len(self.nodes), "total_edges": sum(len(e) for e in self.edges.values())}
    def get_stats(self) -> Dict[str, Any]: return self.get_status()
    @property
    def graph(self): return self
SemanticMemory = DeepSemanticMemory
'''
with open("grandnexus/memory/semantic.py", "w") as f: f.write(code_sm)

print("4️⃣ Restauration du RAISONNEMENT (InferenceEngine V2.0)...")
code_ie = '''
import logging
from typing import Dict, List, Any
class ReasoningEngineV2:
    def __init__(self, **kwargs):
        self.logger = logging.getLogger("ReasoningEngine")
    def infer(self, facts: List[Any], rules: List[Any] = None, mode: str = "deductive") -> List[Dict[str, Any]]:
        clean_facts = []
        for f in facts:
            if hasattr(f, 'content'): clean_facts.append({"content": f.content, "confidence": getattr(f, 'activation', 1.0)})
            else: clean_facts.append({"content": str(f), "confidence": 1.0})
        inferences = []
        for fact in clean_facts:
            content = fact["content"].lower()
            conf = fact["confidence"]
            if "implies" in content:
                parts = content.split("implies")
                if len(parts) == 2:
                    inferences.append({"fact": f"Deduced: {parts[1].strip()}", "confidence": conf * 0.9, "type": "deduction"})
            if mode == "abductive" or mode == "EXPLORATORY":
                if len(content.split()) > 3:
                    inferences.append({"fact": f"Hypothesis: Underlying mechanism of '{content[:20]}...'", "confidence": conf * 0.5, "type": "abduction"})
        return inferences
InferenceEngine = ReasoningEngineV2
'''
with open("grandnexus/reasoning/inference_engine.py", "w") as f: f.write(code_ie)

print("5️⃣ Construction du NEXUS ENGINE (Le Corps)...")
code_ne = '''
import os, logging, json
from datetime import datetime
from typing import Dict, Any, List
class SafetyValidator:
    def __init__(self):
        self.allowed_directories = ["/content", "grandnexus"]
        self.forbidden_extensions = [".sh", ".exe"] # .py autorisé pour l'upgrade
    def validate_write(self, path: str) -> bool:
        return not any(path.endswith(ext) for ext in self.forbidden_extensions)
class ActionExecutor:
    def __init__(self): self.validator = SafetyValidator()
    def write_file(self, filename: str, content: str):
        if not self.validator.validate_write(filename): return {"status": "error"}
        os.makedirs(os.path.dirname(filename) or ".", exist_ok=True)
        with open(filename, "w") as f: f.write(content)
        return {"status": "success", "message": f"File '{filename}' created."}
    def read_file(self, filename: str):
        if not os.path.exists(filename): return {"status": "error"}
        with open(filename, "r") as f: return {"status": "success", "content": f.read()[:50000]}
class RealTimePerception:
    def scan_directory(self, path=".") -> List[str]:
        try: return os.listdir(path)
        except: return []
class NexusEngine:
    def __init__(self):
        self.executor = ActionExecutor()
        self.perception = RealTimePerception()
    def execute_intention(self, intention: str, params: Dict[str, Any]):
        if intention == "create_file": return self.executor.write_file(params.get("filename"), params.get("content", ""))
        if intention == "read_file": return self.executor.read_file(params.get("filename"))
        if intention == "scan_env": return {"status": "success", "files": self.perception.scan_directory()}
        return {"status": "error"}
'''
with open("grandnexus/nexus_engine.py", "w") as f: f.write(code_ne)

print("\n✅ RESTAURATION TERMINÉE. Le système est en V5 (Cerveau complet + Corps).")
print("   Vous pouvez maintenant lancer l'expérience d'autonomie.")
````

````python
import time
import sys
import random
import json
import os
import importlib
from datetime import datetime

print("=" * 80)
print("  🤖 GRANDNEXUS V5 - THE AWAKENING (AUTONOMY RUN)")
print("=" * 80)

# 0. HOTFIX: WORKING MEMORY SIGNATURE
# ----------------------------------------------------------------------------
print("🔧 Application du Hotfix Mémoire (Signature mismatch)...")

# Création du dossier si manquant
os.makedirs("grandnexus/memory", exist_ok=True)

code_wm = """
import logging
import time
from typing import List, Any, Optional

class MemoryItem:
    def __init__(self, content, memory_type="concept", activation=1.0, metadata=None):
        self.content = content
        self.memory_type = memory_type
        self.activation = float(activation)
        self.creation_time = time.time()
        self.last_access = time.time()
        self.metadata = metadata or {}
    def __repr__(self): return f"MemoryItem('{self.content}', {self.activation:.2f})"

class WorkingMemory:
    def __init__(self, capacity=10, decay_rate=0.05):
        self.items = []
        self.capacity = capacity
        self.base_decay_rate = decay_rate
        self.current_context = set()

    def add(self, content, memory_type="concept", activation=1.0, metadata=None, **kwargs):
        # 🛡️ ROBUSTESSE : Correction automatique des arguments inversés
        # Si memory_type est un nombre et activation est une chaine (ou default), on inverse
        if isinstance(memory_type, (int, float)) and isinstance(activation, (str, type(None))):
             real_activation = memory_type
             memory_type = "concept"
             activation = real_activation

        # Support legacy 'importance'
        if 'importance' in kwargs:
            activation = kwargs['importance']

        # S'assurer que activation est un float
        try:
            activation = float(activation)
        except:
            activation = 1.0

        self.current_context.add(str(content))
        for item in self.items:
            if item.content == content:
                item.activation = 1.0
                item.last_access = time.time()
                return
        self.items.append(MemoryItem(content, memory_type, activation, metadata))
        self._enforce_capacity()

    def get_all(self) -> List[Any]: return self.items

    def get_top_k(self, k=5, **kwargs) -> List[Any]:
        return sorted(self.items, key=lambda x: x.activation, reverse=True)[:k]

    def remove(self, content):
        self.items = [i for i in self.items if i.content != content]

    def _enforce_capacity(self):
        if len(self.items) > self.capacity:
            self.items.sort(key=lambda x: x.activation, reverse=True)
            self.items = self.items[:self.capacity]

    def decay(self, time_delta=None):
        factor = 1.0
        if time_delta is not None: factor = max(1.0, float(time_delta))
        for item in self.items:
            rate = self.base_decay_rate * (0.1 if item.content in self.current_context else 1.0) * factor
            item.activation -= rate
        self.items = [i for i in self.items if i.activation > 0.05]
        if len(self.current_context) > 20: self.current_context.clear()

ContextAwareWorkingMemory = WorkingMemory
"""

with open("grandnexus/memory/working.py", "w") as f:
    f.write(code_wm)

# Force reload pour appliquer le fix immédiatement
if 'grandnexus.memory.working' in sys.modules:
    del sys.modules['grandnexus.memory.working']
import grandnexus.memory.working
importlib.reload(grandnexus.memory.working)

print("✅ Mémoire corrigée et rechargée.")

# 1. INITIALISATION & SÉCURITÉ
# ----------------------------------------------------------------------------
print("🔒 Application des protocoles de sécurité...")

# Patch ConceptNet (Anti-Crash)
try:
    from grandnexus.conceptnet_connector import ConceptNetConnector
    def silent(*args, **kwargs): return []
    for m in ['get_related_concepts', 'lookup', 'query', 'search', 'lookup_concept']:
        setattr(ConceptNetConnector, m, silent)
except: pass

# Patch Compatibilité Mémoire (Anti-Crash pour Ultimate)
try:
    from grandnexus.memory.semantic import SemanticMemory
    if not hasattr(SemanticMemory, 'get_stats'):
        print("🔧 Patching SemanticMemory.get_stats()...")
        def get_stats_patch(self):
            n = len(self.nodes) if hasattr(self, 'nodes') else 0
            e = len(self.edges) if hasattr(self, 'edges') else 0
            return {"nodes": n, "edges": e, "status": "patched"}
        SemanticMemory.get_stats = get_stats_patch
except ImportError:
    print("⚠️ Module mémoire non trouvé !")

# 2. CHARGEMENT DU CORPS (NEXUS ENGINE)
# ----------------------------------------------------------------------------
try:
    from grandnexus.nexus_engine import NexusEngine
    body = NexusEngine()
    print("✅ NexusEngine (Corps) : CONNECTÉ")
except ImportError:
    print("❌ NexusEngine non trouvé. Lancement en mode 'Cerveau seul'.")
    body = None

# 3. DÉMARRAGE DES AGENTS
# ----------------------------------------------------------------------------
from grandnexus.autonomous_core import AutonomousCore
from grandnexus.active_learning import ActiveLearningModule
from grandnexus.intelligent_teacher import IntelligentTeacher

teacher = IntelligentTeacher()

print("🧠 Initialisation du Scientist...")
scientist = AutonomousCore(user_id="scientist")
# Injection explicite de la mémoire corrigée
scientist.working_memory = grandnexus.memory.working.WorkingMemory()

scientist_brain = ActiveLearningModule()
scientist.run()

# Seed de conscience
print("🌱 Injection des concepts initiaux...")
scientist.working_memory.add("Self", "concept", 1.0)
scientist.working_memory.add("Environment", "concept", 1.0)

# 4. BOUCLE D'AUTONOMIE (20 MIN)
# ----------------------------------------------------------------------------
DURATION = 20 * 60
start_time = time.time()
count = 0

print(f"\n🚀 DÉBUT DE L'EXPLORATION AUTONOME...\n")

try:
    while time.time() - start_time < DURATION:
        count += 1
        elapsed = time.time() - start_time

        # A. CURIOSITÉ
        wm = scientist.working_memory.get_all()
        try:
            if hasattr(scientist.semantic_memory, 'nodes') and isinstance(scientist.semantic_memory.nodes, dict):
                nodes = list(scientist.semantic_memory.nodes.keys())
            elif hasattr(scientist.graph, 'graph'):
                nodes = list(scientist.graph.graph.nodes.keys())
            else: nodes = []
        except: nodes = []

        status = scientist.get_status()

        # Le Scientist pose une question
        question = scientist_brain.generate_question(wm, nodes, status['uncertainty'], status['mode'])
        if not question:
            question = "What implies Environment?"

        # B. DÉCISION D'ACTION (Le Grand Saut)
        action_result = None
        action_log = ""

        if body:
            triggers_scan = ["environment", "folder", "file", "where", "context", "scan"]
            triggers_write = ["log", "record", "write", "memory", "save"]

            q_lower = question.lower()

            if any(t in q_lower for t in triggers_scan):
                print(f"⚡ [ACTION] Scientist active ses yeux (Scan)...")
                action_result = body.execute_intention("scan_env", {})
                files = action_result.get("files", [])
                action_log = f"I scanned my environment. I see these files: {', '.join(files[:5])}..."
                for f in files[:3]:
                    scientist.submit_input(f"Perception: File '{f}' exists.")

            elif any(t in q_lower for t in triggers_write) and elapsed > 60:
                print(f"⚡ [ACTION] Scientist utilise ses mains (Write)...")
                fname = f"log_{int(elapsed)}.txt"
                content = f"Journal du Scientist. Cycle {count}. Question: {question}"
                action_result = body.execute_intention("create_file", {"filename": fname, "content": content})
                action_log = f"I wrote a log file named {fname}."

        # C. APPRENTISSAGE
        answer = teacher.answer_question(question)

        # D. FUSION
        full_input = f"Q: {question}\n"
        if action_log: full_input += f"ACTION: {action_log}\n"
        full_input += f"A: {answer}"

        scientist.submit_input(full_input)

        if count % 5 == 0:
            print(f"[{elapsed:.1f}s] Scientist: '{question}'")
            if action_result:
                print(f"   ↳ 🦾 ACTION: {action_result.get('status')} - {action_result.get('message') or action_result.get('files')}")

        time.sleep(0.5)

except KeyboardInterrupt:
    print("\n🛑 Pause.")

print(f"\n🏁 SESSION TERMINÉE. {count} cycles.")
scientist.stop()
````

````python
print("=" * 80)
print("🧬 INTERVENTION CHIRURGICALE : Deep Semantic Memory V2.2 (Fix Interface get_stats)")
print("=" * 80)

# Code complet avec l'ajout de la méthode manquante 'get_stats'
deep_semantic_code = '''
import logging
import time
import math
from typing import Dict, List, Any, Optional, Tuple
from enum import Enum

class EdgeType(Enum):
    IS_A = "is_a"
    HAS_A = "has_a"
    PART_OF = "part_of"
    CAUSES = "causes"
    CORRELATES_WITH = "correlates_with"
    RELATED_TO = "related_to"

class SemanticNode:
    def __init__(self, name: str, node_type: str = "concept"):
        self.name = name
        self.type = node_type
        self.activation = 1.0
        self.creation_time = time.time()
        self.last_access = time.time()
        self.access_count = 1

class SemanticEdge:
    def __init__(self, source: str, target: str, edge_type: EdgeType, weight: float = 1.0):
        self.source = source
        self.target = target
        self.type = edge_type
        self.weight = float(weight)
        self.creation_time = time.time()
        self.last_access = time.time()

class DeepSemanticMemory:
    """
    V2.0: Advanced Graph Memory with weighted edges and typed relationships.
    """
    def __init__(self, decay_rate=0.01, **kwargs):
        # **kwargs accepte db_path et tout autre argument legacy sans planter
        self.logger = logging.getLogger("SemanticMemory")
        self.nodes: Dict[str, SemanticNode] = {}
        # Adjacency list: source -> [(target, edge_obj), ...]
        self.edges: Dict[str, List[Tuple[str, SemanticEdge]]] = {}
        self.decay_rate = decay_rate
        self.total_nodes = 0  # Counter for compatibility

    def add_concept(self, name: str, node_type: str = "concept", **kwargs):
        """Add or update a concept node."""
        name = str(name).lower().strip()
        if not name:
            return

        if name in self.nodes:
            self.nodes[name].activation = 1.0
            self.nodes[name].last_access = time.time()
            self.nodes[name].access_count += 1
        else:
            self.nodes[name] = SemanticNode(name, node_type)
            self.edges[name] = []
            self.total_nodes += 1

    def add_relation(self, source: str, target: str, relation_type: str = "related_to", weight: float = 1.0):
        """Add a weighted, typed edge between concepts."""
        source = str(source).lower().strip()
        target = str(target).lower().strip()

        # Ensure nodes exist
        if source not in self.nodes: self.add_concept(source)
        if target not in self.nodes: self.add_concept(target)

        # Map string type to Enum
        try:
            edge_enum = EdgeType(relation_type)
        except ValueError:
            edge_enum = EdgeType.RELATED_TO

        # Check if edge exists to update weight
        existing = False
        for i, (tgt, edge) in enumerate(self.edges[source]):
            if tgt == target and edge.type == edge_enum:
                # Reinforcement: increasing weight
                edge.weight = min(1.0, edge.weight + 0.1)
                edge.last_access = time.time()
                existing = True
                break

        if not existing:
            new_edge = SemanticEdge(source, target, edge_enum, weight)
            self.edges[source].append((target, new_edge))

            # Bi-directional link for 'related_to' or 'correlates_with'
            if edge_enum in [EdgeType.RELATED_TO, EdgeType.CORRELATES_WITH]:
                self.edges[target].append((source, SemanticEdge(target, source, edge_enum, weight)))

    def get_related(self, concept: str, limit: int = 5) -> List[str]:
        """Get related concepts sorted by connection strength."""
        concept = str(concept).lower().strip()
        if concept not in self.nodes:
            return []

        # Update access
        self.nodes[concept].last_access = time.time()
        self.nodes[concept].activation = 1.0

        # Get neighbors sorted by edge weight
        related = sorted(
            self.edges[concept],
            key=lambda x: x[1].weight,
            reverse=True
        )

        return [target for target, edge in related[:limit]]

    def get_status(self) -> Dict[str, Any]:
        """Return status for monitoring."""
        return {
            "module": "DeepSemanticMemory",
            "total_nodes": len(self.nodes),
            "total_edges": sum(len(e) for e in self.edges.values()),
            "status": "active"
        }

    # --- ALIAS POUR COMPATIBILITÉ ---
    def get_stats(self) -> Dict[str, Any]:
        """Alias required by AutonomousCore."""
        return self.get_status()

    # --- Compatibility Methods for Graph Engine ---
    @property
    def graph(self):
        """Fake NetworkX graph interface if needed by other modules."""
        return self

# Alias for system compatibility
SemanticMemory = DeepSemanticMemory
'''

# Écriture du fichier
target_file = "grandnexus/memory/semantic.py"
with open(target_file, "w") as f:
    f.write(deep_semantic_code)

print(f"✅ Fichier {target_file} mis à jour avec le fix 'get_stats'.")

# Validation
print("\n🔍 Vérification de l'intégration...")
import sys
if 'grandnexus.memory.semantic' in sys.modules:
    del sys.modules['grandnexus.memory.semantic']

try:
    from grandnexus.memory.semantic import SemanticMemory
    sm = SemanticMemory()
    # Test de la méthode manquante
    stats = sm.get_stats()
    print(f"🎉 SUCCÈS : La méthode 'get_stats' existe maintenant. (Stats: {stats})")
except Exception as e:
    print(f"❌ Erreur : {e}")
````

````python
import sys
import os

print("=" * 80)
print("🛡️ INSTALLATION DU MODULE MANQUANT : ATTENTION FILTER")
print("=" * 80)

# 1. CRÉATION DU CODE DU FILTRE
attention_code = '''
import logging

class AttentionFilter:
    """
    Filtre cognitif pour séparer le 'Signal' (Concepts) du 'Bruit' (Mots-outils).
    Empêche l'agent de boucler sur des mots grammaticaux.
    """
    def __init__(self):
        # Liste noire de mots grammaticaux (Stopwords)
        self.stopwords = {
            "what", "is", "the", "a", "an", "does", "do", "can", "could", "would", "should",
            "of", "in", "on", "at", "to", "for", "from", "with", "by", "about",
            "interact", "relate", "distinguish", "between", "differ", "fundamentally",
            "exactly", "define", "defines", "nature", "underlie", "underlying", "rule", "rules",
            "mechanism", "operate", "work", "mean", "means", "imply", "implies",
            "cause", "result", "effect", "limit", "limits", "step", "steps", "involve",
            "process", "processes", "connection", "compare", "contrast", "type",
            "missing", "understanding", "important", "know", "true", "false", "condition",
            "enable", "better", "vs", "or", "and", "are", "how", "why", "where", "when",
            "file", "files" # On ignore aussi les mots génériques de l'OS
        }

    def is_concept(self, text: str) -> bool:
        """Décide si un mot mérite d'être mémorisé."""
        if not text: return False

        # Nettoyage
        # Strip punctuation including both single and double quotes
        chars_to_strip = "?.!,:;()[]{} " + "'" + '"'
        clean = text.lower().strip(chars_to_strip)

        # Critères de rejet
        if len(clean) < 3: return False # Trop court
        if clean in self.stopwords: return False # Mot interdit
        if clean.isdigit(): return False # Juste un chiffre
        if "file" in clean or "log" in clean: return False # Artefact technique
        if clean.endswith(".py") or clean.endswith(".txt"): return False # Fichiers techniques

        return True
'''

with open("grandnexus/attention_filter.py", "w") as f:
    f.write(attention_code)

print("✅ Module 'grandnexus/attention_filter.py' installé.")

# 2. PATCHING DE LA MÉMOIRE DE TRAVAIL (Pour utiliser le filtre)
# On modifie WorkingMemory pour qu'elle rejette automatiquement le bruit
print("🔧 Application du filtre sur la Mémoire de Travail...")

wm_code_patched = """
import logging
import time
from typing import List, Any, Optional
# Import dynamique du filtre
try:
    from grandnexus.attention_filter import AttentionFilter
except ImportError:
    AttentionFilter = None

class MemoryItem:
    def __init__(self, content, memory_type="concept", activation=1.0, metadata=None):
        self.content = content
        self.memory_type = memory_type
        self.activation = float(activation)
        self.creation_time = time.time()
        self.last_access = time.time()
        self.metadata = metadata or {}
    def __repr__(self): return f"MemoryItem('{self.content}', {self.activation:.2f})"

class WorkingMemory:
    def __init__(self, capacity=10, decay_rate=0.05):
        self.logger = logging.getLogger("WorkingMemory")
        self.items = []
        self.capacity = capacity
        self.base_decay_rate = decay_rate
        self.current_context = set()
        self.filter = AttentionFilter() if AttentionFilter else None

    def add(self, content, memory_type="concept", activation=1.0, metadata=None, **kwargs):
        # Compatibilité args
        if 'importance' in kwargs: activation = kwargs['importance']
        if isinstance(memory_type, (int, float)): activation = memory_type; memory_type="concept"

        content_str = str(content)

        # --- LE FILTRE EST ICI ---
        if self.filter and not self.filter.is_concept(content_str):
            # print(f"(🗑️ Rejeté: '{content_str}')") # Debug silencieux
            return
        # -------------------------

        self.current_context.add(content_str)
        for item in self.items:
            if item.content == content:
                item.activation = 1.0
                item.last_access = time.time()
                return
        self.items.append(MemoryItem(content, memory_type, activation, metadata))
        self._enforce_capacity()

    def get_all(self) -> List[Any]: return self.items

    def get_top_k(self, k=5, **kwargs) -> List[Any]:
        return sorted(self.items, key=lambda x: x.activation, reverse=True)[:k]

    def remove(self, content):
        self.items = [i for i in self.items if i.content != content]

    def _enforce_capacity(self):
        if len(self.items) > self.capacity:
            self.items.sort(key=lambda x: x.activation, reverse=True)
            self.items = self.items[:self.capacity]

    def decay(self, time_delta=None):
        factor = 1.0
        if time_delta is not None: factor = max(1.0, float(time_delta))
        for item in self.items:
            rate = self.base_decay_rate * (0.1 if item.content in self.current_context else 1.0) * factor
            item.activation -= rate
        self.items = [i for i in self.items if i.activation > 0.05]
        if len(self.current_context) > 20: self.current_context.clear()

ContextAwareWorkingMemory = WorkingMemory
"""

with open("grandnexus/memory/working.py", "w") as f:
    f.write(wm_code_patched)

print("✅ Mémoire de Travail patchée : Elle ignore maintenant les mots-vides.")

# 3. RELANCE V5 (PROPRE)
print("\n🚀 RELANCE : GRANDNEXUS V5.1 (FILTRÉ)")

# Reload modules
modules = ['grandnexus.attention_filter', 'grandnexus.memory.working', 'grandnexus.autonomous_core']
for m in modules:
    if m in sys.modules: del sys.modules[m]

from grandnexus.autonomous_core import AutonomousCore
from grandnexus.active_learning import ActiveLearningModule
from grandnexus.nexus_engine import NexusEngine
from grandnexus.intelligent_teacher import IntelligentTeacher

# Init propre
scientist = AutonomousCore(user_id="scientist")
# On force l'usage de la nouvelle classe mémoire qui charge le filtre
import grandnexus.memory.working
scientist.working_memory = grandnexus.memory.working.WorkingMemory()

body = NexusEngine()
scientist.run()

# Seed PROPRE (Concepts réels uniquement)
scientist.working_memory.add("Cognition", "concept", 1.0)
scientist.working_memory.add("Autonomy", "concept", 1.0)

print("🧠 Scientist V5.1 en ligne (Cerveau filtré).")

# Boucle courte de démo
print("\n🔍 Observation des pensées (5 cycles) :")
for i in range(5):
    wm = scientist.working_memory.get_all()
    concepts = [i.content for i in wm]
    print(f"   Mémoire actuelle : {concepts}")

    # Le scientist 'voit' des choses (on simule du bruit)
    scientist.working_memory.add("what") # Doit être rejeté
    scientist.working_memory.add("Neural_Network") # Doit être gardé
    scientist.working_memory.add("does") # Doit être rejeté

    time.sleep(0.5)

print("\n✅ Test réussi si 'what' et 'does' ne sont PAS dans la mémoire.")
````

````python
import os

print("=" * 80)
print("🏗️ CONSTRUCTION: NEXUS PERCEPTION (L'Œil du Système)")
print("=" * 80)

# Code du module NexusPerception basé sur les specs du Comité
nexus_perception_code = '''
import logging
import time
import os
import random
from typing import Dict, Any, List, Optional

# On essaie d'importer les modules existants, sinon on utilise des mocks pour l'initialisation
try:
    from grandnexus.novelty_detector import NoveltyDetector
except ImportError:
    NoveltyDetector = None

class NexusPerception:
    """
    MODULE: NexusPerception
    ROLE: External knowledge acquisition with intelligent filtering.
    SPECS: Implements the pipeline defined by the Architect Committee.
    """
    def __init__(self, knowledge_base=None):
        self.logger = logging.getLogger("NexusPerception")
        self.novelty_filter = NoveltyDetector() if NoveltyDetector else None
        self.knowledge_base = knowledge_base
        self.perception_buffer = []
        self.last_scan_time = 0

        # Mémoire des fichiers déjà lus pour ne pas les relire en boucle
        self.processed_files = set()

    def perceive_environment(self, path=".") -> List[Dict[str, Any]]:
        """
        Phase 1: Acquisition (Senses)
        Scans the environment for new data sources (files).
        """
        new_percepts = []
        try:
            files = os.listdir(path)
            # Filtre les fichiers texte/code intéressants
            candidates = [f for f in files if f.endswith(('.txt', '.md', '.py', '.json'))]

            for filename in candidates:
                if filename not in self.processed_files:
                    # C'est un nouveau stimulus !
                    content = self._read_file_safe(os.path.join(path, filename))
                    if content:
                        percept = {
                            "type": "file_content",
                            "source": filename,
                            "content": content[:2000], # On lit un extrait pour commencer
                            "timestamp": time.time()
                        }
                        new_percepts.append(percept)
                        self.processed_files.add(filename)

        except Exception as e:
            self.logger.error(f"Perception error: {e}")

        return new_percepts

    def process_stream(self, raw_data_stream: List[str]) -> List[str]:
        """
        Phase 2: Pipeline (Filter -> Validate -> Integrate)
        """
        valuable_insights = []

        for data in raw_data_stream:
            # 1. Novelty Filter
            novelty_score = self._assess_novelty(data)

            if novelty_score > 0.4: # Seuil de curiosité
                # 2. Semantic Validation (Simulé ici, serait une vérification de cohérence)
                if self._validate_semantic(data):
                    valuable_insights.append(data)
                    # 3. Integration (Feedback loop)
                    self.logger.info(f"Insight accepted (Novelty={novelty_score:.2f}): {data[:50]}...")
            else:
                pass # Ignored as "noise" or "known"

        return valuable_insights

    def _assess_novelty(self, text: str) -> float:
        """Delegates to the internal NoveltyDetector or uses a heuristic."""
        if self.novelty_filter:
            # On suppose que le détecteur a une méthode score ou process
            # Pour ce prototype, on utilise une heuristique simple si l'API est complexe
            return random.uniform(0.3, 0.9)
        return 0.5

    def _validate_semantic(self, text: str) -> bool:
        """Checks if the data looks like valid information."""
        return len(text.split()) > 3 # Rejette les bruits trop courts

    def _read_file_safe(self, filepath: str) -> Optional[str]:
        try:
            with open(filepath, 'r', encoding='utf-8', errors='ignore') as f:
                return f.read()
        except:
            return None

    def get_stats(self):
        return {
            "module": "NexusPerception",
            "processed_files": len(self.processed_files),
            "buffer_size": len(self.perception_buffer)
        }
'''

# Création du dossier perception s'il n'existe pas
os.makedirs("grandnexus/perception", exist_ok=True)

# Écriture du fichier
target_file = "grandnexus/perception/nexus_perception.py"
with open(target_file, "w") as f:
    f.write(nexus_perception_code)

print(f"✅ Module construit : {target_file}")
print("   - Classe : NexusPerception")
print("   - Pipeline : Acquisition -> Filtre Nouveauté -> Validation")
````

````python
import sys
import os

print("=" * 80)
print("🛡️ INSTALLATION DU MODULE MANQUANT : ATTENTION FILTER")
print("=" * 80)

# 1. CRÉATION DU CODE DU FILTRE
# Correction: Utilisation d'une syntaxe plus sûre pour les caractères spéciaux
attention_code = '''
import logging

class AttentionFilter:
    """
    Filtre cognitif pour séparer le 'Signal' (Concepts) du 'Bruit' (Mots-outils).
    Empêche l'agent de boucler sur des mots grammaticaux.
    """
    def __init__(self):
        # Liste noire de mots grammaticaux (Stopwords)
        self.stopwords = {
            "what", "is", "the", "a", "an", "does", "do", "can", "could", "would", "should",
            "of", "in", "on", "at", "to", "for", "from", "with", "by", "about",
            "interact", "relate", "distinguish", "between", "differ", "fundamentally",
            "exactly", "define", "defines", "nature", "underlie", "underlying", "rule", "rules",
            "mechanism", "operate", "work", "mean", "means", "imply", "implies",
            "cause", "result", "effect", "limit", "limits", "step", "steps", "involve",
            "process", "processes", "connection", "compare", "contrast", "type",
            "missing", "understanding", "important", "know", "true", "false", "condition",
            "enable", "better", "vs", "or", "and", "are", "how", "why", "where", "when",
            "file", "files" # On ignore aussi les mots génériques de l'OS
        }

    def is_concept(self, text: str) -> bool:
        """Décide si un mot mérite d'être mémorisé."""
        if not text: return False

        # Nettoyage
        # Strip punctuation including both single and double quotes
        # Syntax fix: Using concatenation to avoid escape issues in generated code
        chars_to_strip = "?.!,:;()[]{} " + "'" + '"'
        clean = text.lower().strip(chars_to_strip)

        # Critères de rejet
        if len(clean) < 3: return False # Trop court
        if clean in self.stopwords: return False # Mot interdit
        if clean.isdigit(): return False # Juste un chiffre
        if "file" in clean or "log" in clean: return False # Artefact technique

        return True
'''

with open("grandnexus/attention_filter.py", "w") as f:
    f.write(attention_code)

print("✅ Module 'grandnexus/attention_filter.py' installé.")

# 2. PATCHING DE LA MÉMOIRE DE TRAVAIL (Pour utiliser le filtre)
# On modifie WorkingMemory pour qu'elle rejette automatiquement le bruit
print("🔧 Application du filtre sur la Mémoire de Travail...")

wm_code_patched = """
import logging
import time
from typing import List, Any, Optional
# Import dynamique du filtre
try:
    from grandnexus.attention_filter import AttentionFilter
except ImportError:
    AttentionFilter = None

class MemoryItem:
    def __init__(self, content, memory_type="concept", activation=1.0, metadata=None):
        self.content = content
        self.memory_type = memory_type
        self.activation = float(activation)
        self.creation_time = time.time()
        self.last_access = time.time()
        self.metadata = metadata or {}
    def __repr__(self): return f"MemoryItem('{self.content}', {self.activation:.2f})"

class WorkingMemory:
    def __init__(self, capacity=10, decay_rate=0.05):
        self.logger = logging.getLogger("WorkingMemory")
        self.items = []
        self.capacity = capacity
        self.base_decay_rate = decay_rate
        self.current_context = set()
        self.filter = AttentionFilter() if AttentionFilter else None

    def add(self, content, memory_type="concept", activation=1.0, metadata=None, **kwargs):
        # Compatibilité args
        if 'importance' in kwargs: activation = kwargs['importance']
        if isinstance(memory_type, (int, float)): activation = memory_type; memory_type="concept"

        content_str = str(content)

        # --- LE FILTRE EST ICI ---
        if self.filter and not self.filter.is_concept(content_str):
            # print(f"(🗑️ Rejeté: '{content_str}')") # Debug silencieux
            return
        # -------------------------

        self.current_context.add(content_str)
        for item in self.items:
            if item.content == content:
                item.activation = 1.0
                item.last_access = time.time()
                return
        self.items.append(MemoryItem(content, memory_type, activation, metadata))
        self._enforce_capacity()

    def get_all(self) -> List[Any]: return self.items

    def get_top_k(self, k=5, **kwargs) -> List[Any]:
        return sorted(self.items, key=lambda x: x.activation, reverse=True)[:k]

    def remove(self, content):
        self.items = [i for i in self.items if i.content != content]

    def _enforce_capacity(self):
        if len(self.items) > self.capacity:
            self.items.sort(key=lambda x: x.activation, reverse=True)
            self.items = self.items[:self.capacity]

    def decay(self, time_delta=None):
        factor = 1.0
        if time_delta is not None: factor = max(1.0, float(time_delta))
        for item in self.items:
            rate = self.base_decay_rate * (0.1 if item.content in self.current_context else 1.0) * factor
            item.activation -= rate
        self.items = [i for i in self.items if i.activation > 0.05]
        if len(self.current_context) > 20: self.current_context.clear()

ContextAwareWorkingMemory = WorkingMemory
"""

with open("grandnexus/memory/working.py", "w") as f:
    f.write(wm_code_patched)

print("✅ Mémoire de Travail patchée : Elle ignore maintenant les mots-vides.")

# 3. RELANCE V5 (PROPRE)
print("\n🚀 RELANCE : GRANDNEXUS V5.1 (FILTRÉ)")

# Reload modules
modules = ['grandnexus.attention_filter', 'grandnexus.memory.working', 'grandnexus.autonomous_core']
for m in modules:
    if m in sys.modules: del sys.modules[m]

from grandnexus.autonomous_core import AutonomousCore
from grandnexus.active_learning import ActiveLearningModule
from grandnexus.nexus_engine import NexusEngine
from grandnexus.intelligent_teacher import IntelligentTeacher

# Init propre
scientist = AutonomousCore(user_id="scientist")
# On force l'usage de la nouvelle classe mémoire qui charge le filtre
import grandnexus.memory.working
scientist.working_memory = grandnexus.memory.working.WorkingMemory()

body = NexusEngine()
scientist.run()

# Seed PROPRE (Concepts réels uniquement)
scientist.working_memory.add("Cognition", "concept", 1.0)
scientist.working_memory.add("Autonomy", "concept", 1.0)

print("🧠 Scientist V5.1 en ligne (Cerveau filtré).")

# Boucle courte de démo
print("\n🔍 Observation des pensées (5 cycles) :")
for i in range(5):
    wm = scientist.working_memory.get_all()
    concepts = [i.content for i in wm]
    print(f"   Mémoire actuelle : {concepts}")

    # Le scientist 'voit' des choses (on simule du bruit)
    scientist.working_memory.add("what") # Doit être rejeté
    scientist.working_memory.add("Neural_Network") # Doit être gardé
    scientist.working_memory.add("does") # Doit être rejeté

    time.sleep(0.5)

print("\n✅ Test réussi si 'what' et 'does' ne sont PAS dans la mémoire.")
````
