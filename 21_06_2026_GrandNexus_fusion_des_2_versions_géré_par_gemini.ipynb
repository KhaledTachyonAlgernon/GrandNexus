### 🚀 Apprentissage Intensif et Parallèle
Cette version utilise le multi-threading pour interroger plusieurs sources en même temps.

```python
import concurrent.futures

def worker_wikipedia():
    try:
        page = wikipedia.random()
        content = wikipedia.summary(page, sentences=3)
        return f"[Wiki-Parallel] {page}"
    except:
        return None

def worker_arxiv():
    try:
        search = arxiv.Search(query="AI machine learning", max_results=1)
        paper = next(arxiv_client.results(search))
        return f"[Arxiv-Parallel] {paper.title}"
    except:
        return None

def run_intensive_learning(duration_minutes=5, max_workers=5):
    end_time = time.time() + duration_minutes * 60
    print(f"Mode INTENSIF activé : {max_workers} threads en parallèle.")

    with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
        while time.time() < end_time:
            # On lance un mix de tâches en parallèle
            futures = []
            for _ in range(max_workers // 2):
                futures.append(executor.submit(worker_wikipedia))
                futures.append(executor.submit(worker_arxiv))

            for future in concurrent.futures.as_completed(futures):
                result = future.result()
                if result:
                    print(f"Appris : {result}")

            time.sleep(2) # Petite pause pour respecter les APIs

# Lancer un test intensif de 5 minutes
# run_intensive_learning(duration_minutes=5, max_workers=8)
```

### 🌉 Fusion : Apprentissage Continu + GrandNexusCore
Ici, nous simulons un agent qui lit du texte, en extrait des connaissances structurées (triplets), et les intègre au graphe sémantique pur de `GrandNexusCore`.

```python
import re
from grandnexus_core import GrandNexusCore

# 1. Fonction d'extraction de triplets (Heuristique simplifiée pour la démo)
def extract_triplets(text):
    """
    Extrait des triplets (sujet, relation, objet) d'un texte simple.
    Dans une vraie version, on utiliserait spaCy, un LLM (comme Gemini), ou OpenIE.
    """
    triplets = []
    text = text.lower()

    # Règle très simple : "X est un Y" -> (X, IS_A, Y)
    is_a_pattern = re.compile(r'([\w\s]+) est un ([\w\s]+)[\.,]?')
    for match in is_a_pattern.finditer(text):
        sujet = match.group(1).strip()
        objet = match.group(2).strip()
        triplets.append((sujet, 'IS_A', objet))

    # Règle : "X contient Y" -> (X, HAS_PART, Y)
    has_part_pattern = re.compile(r'([\w\s]+) contient ([\w\s]+)[\.,]?')
    for match in has_part_pattern.finditer(text):
        sujet = match.group(1).strip()
        objet = match.group(2).strip()
        triplets.append((sujet, 'HAS_PART', objet))

    return triplets

# Test de l'extracteur
snippet = "Le cerveau est un organe complexe. Le crâne contient le cerveau."
print("Texte source :", snippet)
print("Triplets extraits :", extract_triplets(snippet))
```

```python
def get_graph(nexus_engine):
    # Cherche dynamiquement l'attribut qui contient le KnowledgeGraph
    for k, v in vars(nexus_engine).items():
        if type(v).__name__ == 'KnowledgeGraph':
            return v
    return None

def inject_knowledge(nexus_engine, triplets):
    """
    Injecte des triplets dans le graphe de connaissances de GrandNexusCore.
    """
    count = 0
    graph = get_graph(nexus_engine)
    if not graph:
        print("Erreur : Impossible de trouver le graphe dans le noyau.")
        return 0

    for sujet, relation, objet in triplets:
        try:
            graph.add(sujet, objet, relation)
            count += 1
            print(f"[Injection] {sujet} --({relation})--> {objet}")
        except Exception as e:
            print(f"Erreur d'injection pour {sujet}-{relation}-{objet}: {e}")
    return count

# On crée une nouvelle instance propre
nexus_fusion = GrandNexusCore()

print("\n=== Injection de Nouvelles Connaissances ===")
new_triplets = extract_triplets(snippet)
inject_knowledge(nexus_fusion, new_triplets)
graph = get_graph(nexus_fusion)
if graph:
    print(f"Taille du graphe après injection : {graph.count()} relations.")
```

```python
class GrandNexusAgent:
    """
    Un agent qui lit des snippets, extrait des faits, et interroge le noyau
    pour valider ou intégrer de nouvelles connaissances.
    """
    def __init__(self, core_engine):
        self.core = core_engine
        self.graph = get_graph(core_engine)

    def process_snippet(self, text, source="Inconnue"):
        print(f"\n[Agent] Lecture depuis {source} : '{text}'")
        triplets = extract_triplets(text)

        if not triplets:
            print("[Agent] Aucun fait structuré extrait.")
            return

        if not self.graph:
            print("Erreur : Graphe inaccessible.")
            return

        for sujet, relation, objet in triplets:
            print(f"[Agent] Hypothèse extraite : {sujet} est lié à {objet} par {relation}")

            # On vérifie si le noyau sait déjà cela (raisonnement IS_A par exemple)
            if relation == 'IS_A':
                check = self.core.is_a(sujet, objet)
                if check.get('reponse') == True:
                    print(f"  -> Le noyau confirme (Déjà su) : {check.get('base')}")
                else:
                    print(f"  -> Le noyau ne connaissait pas ce fait. Injection...")
                    self.graph.add(sujet, objet, relation)
            else:
                # Pour d'autres relations, on injecte directement comme nouvelle connaissance
                print(f"  -> Injection de la relation {relation}...")
                self.graph.add(sujet, objet, relation)

# Démonstration de l'agent
agent = GrandNexusAgent(nexus_fusion)

# 1. Un fait nouveau
agent.process_snippet("Le coeur est un organe vital.", source="Wikipedia")

# 2. Un fait que nous venons juste d'injecter
agent.process_snippet("Le cerveau est un organe complexe.", source="Wikipedia")
```

```python
import concurrent.futures
import threading
import random
import time
import wikipedia
import arxiv

# Verrou pour protéger l'écriture dans le graphe depuis plusieurs threads simultanés
graph_lock = threading.Lock()

def thread_worker(agent, source_name):
    """Worker qui va chercher de l'info, l'analyser et l'injecter de manière autonome."""
    try:
        if source_name == "wikipedia":
            page = wikipedia.random()
            text = wikipedia.summary(page, sentences=3)
            prefix = f"[Wiki: {page[:15]}...]"
        else:
            # Arxiv au hasard sur des thèmes scientifiques
            queries = ["machine learning", "quantum computing", "biology", "robotics"]
            search = arxiv.Search(query=random.choice(queries), max_results=1)
            paper = next(arxiv_client.results(search))
            text = paper.summary.replace('\n', ' ')
            prefix = f"[Arxiv: {paper.title[:15]}...]"

        # 1. Extraction (NLP simple, hors du verrou pour maximiser le parallélisme)
        triplets = extract_triplets(text)

        # 2. Injection (Dans le verrou pour sécuriser le graphe)
        if triplets:
            with graph_lock:
                for sujet, relation, objet in triplets:
                    agent.graph.add(sujet, objet, relation)
            return f"🧠 {prefix} Appris : {triplets}"
        else:
            # On retourne None silencieusement si rien n'a été extrait pour ne pas polluer l'écran
            return None

    except Exception as e:
        return None

def run_real_time_learning(agent, duration_minutes=5, max_threads=4):
    print(f"🚀 Lancement de l'apprentissage auto-supervisé ({duration_minutes} min, {max_threads} cœurs)...")
    end_time = time.time() + (duration_minutes * 60)

    with concurrent.futures.ThreadPoolExecutor(max_workers=max_threads) as executor:
        while time.time() < end_time:
            futures = []
            # On lance une salve de requêtes simultanées
            for _ in range(max_threads):
                # On favorise Wikipedia car notre extracteur de triplets actuel (regex 'est un') y marche mieux
                source = random.choice(["wikipedia", "wikipedia", "arxiv"])
                futures.append(executor.submit(thread_worker, agent, source))

            # On récupère les résultats de la salve
            for future in concurrent.futures.as_completed(futures):
                res = future.result()
                if res:
                    print(res) # Affiche uniquement les succès

            time.sleep(1.5) # Courte pause pour éviter le blocage des API (Rate Limit)

    print("\n✅ Session d'apprentissage terminée !")

# --- INITIALISATION D'UN NOUVEAU NOYAU POUR LE TEST ---
live_core = GrandNexusCore()
live_agent = GrandNexusAgent(live_core)
print("Nouveau noyau instancié. Prêt pour le test.")
```

```python
# -*- coding: utf-8 -*-
"""
GrandNexus-core  —  moteur de connaissances graphe-natif (AUCUN LLM au coeur)
=============================================================================
L'integration des briques *validees une par une* :
  - graphe a ARETES TYPEES (l'essence) ;
  - EXPLORATION autonome d'un domaine (expansion bornee, deterministe, inspectable) ;
  - RAISONNEMENT avec JUSTIFICATION (deduction -> chemin de preuve, pas un recit) ;
  - PREDICTION analogique (methode mesuree a ~18x le hasard) avec score de confiance ;
  - INTERFACE de requete STRUCTUREE -> consommable par un humain OU une autre IA.

Honnetete inscrite dans le code :
  * une reponse "deduite" porte sa preuve (le chemin) ; rien n'est affirme sans trace.
  * une "proposition" est une HYPOTHESE (score heuristique non calibre), jamais un fait.
  * l'"autonomie" = expansion + proposition sans pilotage pas-a-pas ; PAS d'agentivite,
    pas de but propre, pas de conscience. Tout est borne et examinable.

Substrat : WordNet (hors-ligne, fiable). L'enrichissement ConceptNet local et le banc
d'evaluation (cf. noyaux v3-v5) se branchent par-dessus comme couche "instrument de
recherche". Le LLM, lui, ne reviendra qu'en entree/sortie (verbaliser), jamais au coeur.
"""

from collections import defaultdict, deque

try:
    from nltk.corpus import wordnet as wn
    _ = wn.synsets("dog")
except Exception:
    import nltk
    nltk.download("wordnet", quiet=True); nltk.download("omw-1.4", quiet=True)
    from nltk.corpus import wordnet as wn

SYNSET_RELS = {"hypernyms": "IS_A", "part_meronyms": "HAS_PART", "part_holonyms": "PART_OF",
               "member_meronyms": "HAS_MEMBER", "member_holonyms": "MEMBER_OF",
               "substance_meronyms": "MADE_OF", "substance_holonyms": "SUBSTANCE_OF"}


class KnowledgeGraph:
    def __init__(self):
        self.edges = defaultdict(set); self.nodes = set()
    def add(self, s, rel, t):
        if s != t:
            self.nodes.add(s); self.nodes.add(t); self.edges[rel].add((s, t))
    def count(self, rel=None):
        return len(self.edges[rel]) if rel else sum(len(v) for v in self.edges.values())
    def out(self, s, rel):
        return {t for (a, t) in self.edges[rel] if a == s}
    def index(self, rel):
        idx = defaultdict(set)
        for (a, t) in self.edges[rel]:
            idx[a].add(t)
        return idx
    def neighborhood(self, n):
        f = set()
        for rel, pairs in self.edges.items():
            for (a, t) in pairs:
                if a == n: f.add((rel, "o", t))
                elif t == n: f.add((rel, "i", a))
        return f


class GrandNexusCore:
    def __init__(self):
        self.g = KnowledgeGraph()
        self._anc = None
        self._nb = {}

    # ---------- petites aides ----------
    def _syn(self, x):
        if hasattr(x, "name"):
            return x
        if "." in x:
            try:
                return wn.synset(x)
            except Exception:
                return None
        cands = wn.synsets(x, pos="n")
        if not cands:
            return None
        if self.g.nodes:                       # desambiguisation par le CONTEXTE du graphe explore
            in_g = [c for c in cands if c.name() in self.g.nodes]
            if in_g:                            # garder le sens le plus connecte dans le domaine
                return max(in_g, key=lambda c: len(self.g.neighborhood(c.name())))
        return cands[0]   # sinon sens le plus frequent ; le LLM-au-bord tranchera les cas vraiment ambigus
    def _pretty(self, name):
        return name.split(".")[0].replace("_", " ")
    def _ingest(self, syn):
        nm = lambda s: s.name()
        for meth, rel in SYNSET_RELS.items():
            for t in getattr(syn, meth)():
                self.g.add(nm(syn), rel, nm(t))

    # ---------- EXPLORATION autonome (bornee) ----------
    def explore(self, seed, depth=3, max_nodes=1500):
        start = self._syn(seed)
        if start is None:
            return {"erreur": f"'{seed}' introuvable dans WordNet"}
        q, seen, n = deque([(start, 0)]), {start}, 0
        while q and n < max_nodes:
            syn, d = q.popleft()
            self._ingest(syn); n += 1
            if d < depth:
                for nb in (syn.hypernyms() + syn.hyponyms() + syn.part_meronyms()
                           + syn.part_holonyms() + syn.member_holonyms()):
                    if nb not in seen:
                        seen.add(nb); q.append((nb, d + 1))
        self._anc, self._nb = None, {}     # invalider les caches
        return {"germe": start.name(), "synsets_ingeres": n,
                "concepts": len(self.g.nodes), "aretes": self.g.count()}

    # ---------- RAISONNEMENT + justification ----------
    def _ancestors(self):
        if self._anc is None:
            idx = self.g.index("IS_A"); C = {}
            def anc(x):
                if x in C: return C[x]
                C[x] = set(); r = set()
                for p in idx.get(x, ()):
                    r.add(p); r |= anc(p)
                C[x] = r; return r
            for x in list(idx): anc(x)
            self._anc = C
        return self._anc
    def _path(self, s, t):
        idx = self.g.index("IS_A"); prev = {s: None}; q = deque([s])
        while q:
            x = q.popleft()
            for p in idx.get(x, ()):
                if p not in prev:
                    prev[p] = x
                    if p == t:
                        path = [t]
                        while prev[path[-1]] is not None: path.append(prev[path[-1]])
                        return path[::-1]
                    q.append(p)
        return None

    def is_a(self, x, y):
        sx, sy = self._syn(x), self._syn(y)
        if not sx or not sy:
            return {"reponse": None, "base": "concept inconnu"}
        a, b = sx.name(), sy.name()
        if (a, b) in self.g.edges["IS_A"]:
            return {"reponse": True, "base": "ingere", "preuve": [self._pretty(a), self._pretty(b)]}
        if b in self._ancestors().get(a, ()):
            p = self._path(a, b)
            return {"reponse": True, "base": "deduit",
                    "preuve": [self._pretty(z) for z in p] if p else None}
        return {"reponse": False, "base": "non etabli dans le graphe explore"}

    def parts_of(self, x):
        s = self._syn(x)
        if not s: return {"erreur": "concept inconnu"}
        name = s.name(); hp = self.g.index("HAS_PART"); out = []
        for p in hp.get(name, ()):
            out.append({"partie": self._pretty(p), "via": "ingere"})
        for anc in self._ancestors().get(name, ()):
            for p in hp.get(anc, ()):
                out.append({"partie": self._pretty(p), "via": f"herite de '{self._pretty(anc)}'"})
        return {"concept": self._pretty(name), "parties": out}

    # ---------- PREDICTION analogique + confiance + justification ----------
    def _feats(self, n):
        if n not in self._nb:
            self._nb[n] = self.g.neighborhood(n) | {("anc", "*", a) for a in self._ancestors().get(n, ())}
        return self._nb[n]
    @staticmethod
    def _jac(a, b):
        if not a or not b: return 0.0
        i = len(a & b); return i / (len(a) + len(b) - i)

    def propose_links(self, x, relation="HAS_PART", top_k=5):
        s = self._syn(x)
        if not s: return {"erreur": "concept inconnu"}
        name = s.name()
        src = defaultdict(set)
        for (w, p) in self.g.edges[relation]:
            src[p].add(w)
        known = set(self.g.out(name, relation))
        for anc in self._ancestors().get(name, ()):
            known |= self.g.out(anc, relation)
        fx = self._feats(name); scored = []
        for p, owners in src.items():
            if p == name or p in known: continue
            sc = sum(self._jac(fx, self._feats(o)) for o in owners)
            if sc > 0:
                because = sorted(owners, key=lambda o: self._jac(fx, self._feats(o)), reverse=True)[:3]
                scored.append((sc, p, because))
        scored.sort(reverse=True); scored = scored[:top_k]
        mx = scored[0][0] if scored else 1.0
        return {"concept": self._pretty(name),
                "propositions": [{"relation": relation, "cible": self._pretty(p),
                                  "confiance": round(sc / mx, 2),   # heuristique, NON calibree
                                  "parce_que": [self._pretty(o) for o in bc]}
                                 for sc, p, bc in scored]}

    # ---------- PONT : connexion STRUCTURELLE prouvee entre deux concepts ----------
    def _shortest_typed_path(self, a, b):
        if a == b:
            return []
        adj = defaultdict(list)
        for rel, pairs in self.g.edges.items():
            for (s, t) in pairs:
                adj[s].append((rel, "->", t)); adj[t].append((rel, "<-", s))
        prev = {a: None}; q = deque([a])
        while q:
            x = q.popleft()
            for (rel, dr, y) in adj[x]:
                if y not in prev:
                    prev[y] = (x, rel, dr)
                    if y == b:
                        steps = []; cur = b
                        while prev[cur] is not None:
                            px, rl, d = prev[cur]; steps.append((px, rl, d, cur)); cur = px
                        return steps[::-1]
                    q.append(y)
        return None

    def pont(self, a, b):
        sa, sb = self._syn(a), self._syn(b)
        if not sa or not sb:
            return {"erreur": "concept inconnu"}
        A, B = sa.name(), sb.name()
        anc = self._ancestors()
        ca, cb = anc.get(A, set()) | {A}, anc.get(B, set()) | {B}
        common = ca & cb
        out = {"de": self._pretty(A), "vers": self._pretty(B)}
        if common:
            lca = max(common, key=lambda c: len(anc.get(c, set())))
            pa = [A] if A == lca else (self._path(A, lca) or [A])
            pb = [B] if B == lca else (self._path(B, lca) or [B])
            out["ancetre_commun"] = self._pretty(lca)
            out["preuve"] = {"depuis_" + self._pretty(A): [self._pretty(z) for z in pa],
                             "depuis_" + self._pretty(B): [self._pretty(z) for z in pb]}
        else:
            out["ancetre_commun"] = None
        sp = self._shortest_typed_path(A, B)
        if sp:
            out["chemin_le_plus_court"] = [
                (f"{self._pretty(s)}  —{rl}→  {self._pretty(t)}" if d == "->"
                 else f"{self._pretty(s)}  ←{rl}—  {self._pretty(t)}") for (s, rl, d, t) in sp]
        return out

    # ---------- INTERFACE STRUCTUREE (humain OU autre IA) ----------
    def query(self, q):
        t = q.get("type")
        if t == "is_a":        return self.is_a(q["x"], q["y"])
        if t == "parts_of":    return self.parts_of(q["x"])
        if t == "propose":     return self.propose_links(q["x"], q.get("relation", "HAS_PART"), q.get("top_k", 5))
        if t == "pont":        return self.pont(q["a"], q["b"])
        if t == "explore":     return self.explore(q["seed"], q.get("depth", 3), q.get("max_nodes", 1500))
        return {"erreur": f"type de requete inconnu: {t}"}


# ================================== DEMO ====================================
def _show(title, obj):
    import json
    print(f"\n>>> {title}")
    print(json.dumps(obj, ensure_ascii=False, indent=2))

if __name__ == "__main__":
    print("=" * 74); print("GrandNexus-core — demonstration"); print("=" * 74)
    gn = GrandNexusCore()

    _show("EXPLORE le domaine 'vehicle' (autonome, borne)", gn.explore("vehicle", depth=4, max_nodes=1500))

    _show("REQUETE  is_a(car, vehicle)  [avec preuve]", gn.query({"type": "is_a", "x": "car", "y": "vehicle"}))
    _show("REQUETE  is_a(car, animal)   [doit etre faux]", gn.query({"type": "is_a", "x": "car", "y": "animal"}))
    _show("REQUETE  parts_of(car)       [ingere + herite]", gn.query({"type": "parts_of", "x": "car"}))
    _show("REQUETE  propose(motorcycle) [hypotheses + confiance + justification]",
          gn.query({"type": "propose", "x": "motorcycle", "top_k": 5}))

    print("\n" + "-" * 74)
    print("Tout est local, inspectable, sans LLM. Les 'propositions' sont des hypotheses")
    print("(score heuristique), a valider — pas des faits. Couche suivante : enrichissement")
    print("ConceptNet local + banc d'eval (v3-v5) comme instrument de recherche ; LLM en")
    print("entree/sortie pour verbaliser, jamais au coeur.")
```

### 🧠 Développement de l'Intelligence : Inférence Automatique (Raisonnement Déductif)
On ajoute au noyau la capacité de déduire de nouveaux faits sans avoir besoin de les lire, simplement en analysant la logique de ce qu'il sait déjà.

```python
def auto_infer(self):
    """
    Parcourt le graphe pour déduire de nouvelles relations logiques.
    Règle 1 (Transitivité IS_A) : Si A IS_A B et B IS_A C, alors A IS_A C.
    Règle 2 (Héritage HAS_PART) : Si A IS_A B et B HAS_PART C, alors A HAS_PART C.
    """
    new_knowledge = []
    is_a_edges = list(self.g.edges.get('IS_A', set()))
    has_part_edges = list(self.g.edges.get('HAS_PART', set()))

    # Règle 1 : Transitivité
    for a, b in is_a_edges:
        for b2, c in is_a_edges:
            if b == b2 and a != c:
                # Vérifier si on ne le sait pas déjà
                if (a, c) not in self.g.edges.get('IS_A', set()):
                    new_knowledge.append((a, 'IS_A', c, f"Déduit : {a}->{b}->{c}"))

    # Règle 2 : Héritage des propriétés
    for a, b in is_a_edges:
        for b2, c in has_part_edges:
            if b == b2 and a != c:
                if (a, c) not in self.g.edges.get('HAS_PART', set()):
                    new_knowledge.append((a, 'HAS_PART', c, f"Hérité : {a} est un {b} qui contient {c}"))

    # Injection des nouvelles connaissances
    for sujet, relation, objet, justification in new_knowledge:
        self.g.add(sujet, objet, relation)

    return new_knowledge

# Patch dynamique du noyau
GrandNexusCore.auto_infer = auto_infer
print("Nouvelle brique cognitive installée : [Inférence Automatique]")
```

```python
def run_growth_and_sleep_cycle(agent, cycles=3, read_minutes=2):
    """
    Alterne entre des phases de lecture (extraction web) et des phases de 'sommeil'
    où le système consolide et déduit de nouvelles informations par lui-même.
    """
    print(f"🌍 Début du cycle de croissance : {cycles} itérations.")

    for i in range(cycles):
        print(f"\n--- CYCLE {i+1}/{cycles} ---")
        print("📖 Phase d'éveil : Lecture et extraction de données...")
        # On réutilise le worker avancé pour lire
        run_advanced_learning(agent, duration_minutes=read_minutes, max_threads=6)

        print("\n💤 Phase de sommeil : Consolidation et Inférence logicielle...")
        deductions = agent.core.auto_infer()
        print(f"🧠 Pendant son 'sommeil', l'IA a déduit {len(deductions)} nouvelles connaissances par elle-même !")

        if deductions:
            # Afficher quelques exemples de ce qu'il a compris tout seul
            for d in deductions[:5]:
                print(f"  -> {d[3]} (Donc {d[0]} {d[1]} {d[2]})")

    print(f"\n✅ Croissance terminée. Taille totale du graphe : {agent.graph.count()} relations.")

# On lance un cycle complet (Lecture puis Inférence)
# Attention, cela prendra environ 6 minutes (3 cycles de 2 minutes de lecture)
run_growth_and_sleep_cycle(advanced_agent, cycles=3, read_minutes=2)
```

```python
import spacy
import wikipedia
from nltk.corpus import wordnet as wn

# 1. On passe Wikipedia en anglais pour un mapping direct avec l'ontologie WordNet
wikipedia.set_lang("en")

# 2. Chargement du modèle linguistique
print("Chargement du modèle NLP spaCy...")
nlp = spacy.load("en_core_web_sm")

def ground_to_concept(word):
    """
    Vérifie si le mot existe dans l'ontologie de base (WordNet).
    C'est ce qui garantit qu'on manipule des concepts, pas juste du texte aveugle.
    """
    syns = wn.synsets(word, pos="n")
    if syns:
        return syns[0].name() # ID ontologique exact (ex: 'intelligence.n.01')
    return word.lower().strip() # Concept nouveau, non reconnu mais accepté

def extract_triplets_ontological(text):
    """
    Extracteur strict : utilise l'arbre de dépendance (Sujet -> Verbe -> Objet)
    """
    doc = nlp(text)
    triplets = []

    for sent in doc.sents:
        sujet, objet, relation = None, None, None

        # Analyse syntaxique stricte
        for token in sent:
            # Recherche du sujet
            if token.dep_ in ("nsubj", "nsubjpass"):
                sujet = token.lemma_
            # Recherche de l'objet (relié au verbe)
            if token.dep_ in ("attr", "dobj", "pobj"):
                if token.head.lemma_ in ("be", "become", "represent"):
                    objet = token.lemma_
                    relation = "IS_A"
                elif token.head.lemma_ in ("include", "contain", "comprise", "have"):
                    objet = token.lemma_
                    relation = "HAS_PART"

        if sujet and objet and relation:
            # Ancrage ontologique
            sujet_ancrage = ground_to_concept(sujet)
            objet_ancrage = ground_to_concept(objet)

            # Rejet strict des auto-références pour garder un graphe sain
            if sujet_ancrage != objet_ancrage:
                triplets.append((sujet_ancrage, relation, objet_ancrage))

    return triplets

print("Extracteur NLP strict et ontologique initialisé avec succès.")
```

### 🛡️ Priorité 1 : Esprit Critique et Détecteur de Contradictions
Ce module empêche le système d'apprendre des absurdités en vérifiant les racines ontologiques (ex: on ne peut pas être à la fois un animal et une plante).

```python
def detect_contradiction(self, sujet, objet, relation="IS_A"):
    """
    Module d'Esprit Critique : Vérifie si la nouvelle information est absurde
    en se basant sur l'ontologie fondamentale.
    """
    if relation != "IS_A":
        return False, "" # On se concentre sur les contradictions de classification (IS_A)

    try:
        from nltk.corpus import wordnet as wn
        syn_s = wn.synsets(sujet.split('.')[0], pos="n")
        syn_o = wn.synsets(objet.split('.')[0], pos="n")

        if syn_s and syn_o:
            ss = syn_s[0]
            so = syn_o[0]

            # Remonter toute la chaîne des hypernymes pour trouver les "racines" existentielles
            paths_s = list(ss.closure(lambda s: s.hypernyms()))
            paths_o = list(so.closure(lambda s: s.hypernyms()))

            # Définition des catégories fondamentales mutuellement exclusives
            racines_exclusives = ['animal.n.01', 'plant.n.02', 'artifact.n.01', 'person.n.01', 'substance.n.01', 'location.n.01']

            roots_s = set([p.name() for p in paths_s if p.name() in racines_exclusives])
            roots_o = set([p.name() for p in paths_o if p.name() in racines_exclusives])

            # Si le sujet et l'objet appartiennent à des branches fondamentales différentes, c'est une contradiction !
            if roots_s and roots_o and not roots_s.intersection(roots_o):
                cat_s = list(roots_s)[0].split('.')[0].upper()
                cat_o = list(roots_o)[0].split('.')[0].upper()
                return True, f"Contradiction ontologique stricte : '{sujet}' (Famille: {cat_s}) ne peut pas être classé comme '{objet}' (Famille: {cat_o})."

    except Exception as e:
        pass

    return False, ""

# Ajout dynamique de la fonction au noyau
GrandNexusCore.detect_contradiction = detect_contradiction
print("🛡️ Module 'Esprit Critique' activé dans GrandNexusCore.")

# --- TESTS DE L'ESPRIT CRITIQUE ---
print("\n=== Tests de Détection de Contradictions ===")
tests = [
    ("cat", "plant"),         # Chat est une plante (Faux)
    ("dog", "artifact"),      # Chien est un objet (Faux)
    ("car", "vehicle"),       # Voiture est un véhicule (Vrai)
    ("human", "substance")    # Humain est une substance (Faux)
]

for s, o in tests:
    is_contra, raison = strict_core.detect_contradiction(s, o, "IS_A")
    if is_contra:
        print(f"🚫 REJETÉ : '{s}' n'est pas un '{o}'. Raison : {raison}")
    else:
        print(f"✅ ACCEPTÉ : '{s}' pourrait potentiellement être un '{o}'.")
```

### 🧠 Priorité 2 : Méta-Mémoire et Consolidation à Long Terme
Le système s'observe, calcule la force de ses propres concepts, nettoie le bruit et sauvegarde sa « Core Memory » dans un fichier persistant.

```python
import json
import os

def meta_memory_consolidation(self, filepath="grand_nexus_long_term_memory.json"):
    """
    Méta-cognition : Le système évalue ses connaissances, oublie les concepts isolés
    (mémoire à court terme / bruit) et consolide le reste dans une mémoire persistante.
    """
    print("\n🧘‍♂️ Début de l'introspection et de la consolidation de la mémoire...")

    nodes = list(self.g.nodes)

    # Étape 1: Auto-évaluation (Calcul de la 'force' de chaque concept par son degré de connectivité)
    concept_strength = {n: len(self.g.neighborhood(n)) for n in nodes}

    oubli_count = 0
    concepts_conserves = []
    memoire_structuree = {}

    # Étape 2: Filtrage et élagage (Pruning)
    for n, strength in concept_strength.items():
        # Règle de méta-cognition : Si un concept a moins de 2 connexions ET n'est pas un concept fondamental (.n.01)
        if strength < 2 and not str(n).endswith('.n.01'):
            oubli_count += 1
        else:
            concepts_conserves.append(n)
            # On sauvegarde son voisinage pour la mémoire longue
            memoire_structuree[n] = {
                "force": strength,
                "voisinage": list(self.g.neighborhood(n))[:10] # On garde un échantillon des connexions
            }

    print(f"🗑️ Méta-cognition : Identification et suppression de {oubli_count} concepts jugés comme du 'bruit' (incertains ou isolés).")
    print(f"💎 Mémoire à Long Terme : Conservation de {len(concepts_conserves)} concepts forts et interconnectés.")

    # Étape 3: Sauvegarde persistance (Export vers le disque, façon MEGA F____)
    data = {
        "meta_cognition_stats": {
            "total_concepts_evalues": len(nodes),
            "concepts_retenus": len(concepts_conserves),
            "bruit_oublie": oubli_count
        },
        "core_memory": memoire_structuree
    }

    with open(filepath, 'w', encoding='utf-8') as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

    taille_ko = os.path.getsize(filepath) / 1024
    print(f"📁 Consolidation terminée. Mémoire sauvegardée dans '{filepath}'.")
    print(f"⚖️ Taille du fichier généré : {taille_ko:.2f} Ko.")

    return data

# Ajout dynamique au noyau
GrandNexusCore.meta_memory_consolidation = meta_memory_consolidation

# Lancement du cycle de Méta-Cognition sur l'agent avancé (qui possède un grand graphe)
print("\n=== Cycle de Sommeil Profond (Agent Avancé) ===")
stats_memoire = advanced_agent.core.meta_memory_consolidation(filepath="grand_nexus_mega_memory.json")
```

### 🧬 Refonte NLP : Ancrage Ontologique Hiérarchique & Requête Multi-Domaines
L'extracteur va maintenant lier chaque mot extrait à sa racine absolue dans l'arbre des connaissances, empêchant les concepts flottants.

```python
def deep_ontological_mapping(self, concept_name):
    """
    Mappe un concept de manière exhaustive jusqu'à la racine de l'ontologie ('entity.n.01').
    Cela force le système à savoir exactement OÙ se situe la nouvelle connaissance.
    """
    try:
        from nltk.corpus import wordnet as wn
        syns = wn.synsets(concept_name.split('.')[0], pos="n")
        if not syns:
            return False

        syn = syns[0]
        paths = syn.hypernym_paths()
        if paths:
            # Prendre le chemin le plus long jusqu'à la racine
            longest_path = max(paths, key=len)
            for i in range(len(longest_path) - 1):
                child = longest_path[i+1].name()
                parent = longest_path[i].name()
                # Injecter la hiérarchie ascendante dans le graphe
                self.g.add(child, parent, "IS_A")
            return True
    except Exception as e:
        pass
    return False

# Patch du noyau
GrandNexusCore.deep_ontological_mapping = deep_ontological_mapping

print("🌐 Module d'Ancrage Ontologique Profond activé.")

# --- TEST : REQUÊTE MULTI-DOMAINES (DISPARATE DOMAINS) ---
print("\n=== TEST : CONNEXION MULTI-DOMAINES (Biologie ↔ Informatique) ===")

# 1. On ancre profondément deux domaines très différents
print("1. Cartographie hiérarchique des domaines...")
advanced_agent.core.deep_ontological_mapping("neuron")    # Biologie / Neuroscience
advanced_agent.core.deep_ontological_mapping("microprocessor") # Technologie / Informatique

# On s'assure que les nœuds locaux sont bien connectés (en explorant un peu autour)
advanced_agent.core.explore("neuron", depth=2)
advanced_agent.core.explore("microprocessor", depth=2)

# 2. Requête multi-sauts complexe (Le Pont)
print("\n2. Exécution de la requête relationnelle complexe...")
multi_step_query = advanced_agent.core.pont("neuron", "microprocessor")

if multi_step_query.get('ancetre_commun'):
    print(f"Ancêtre fondamental commun trouvé : {multi_step_query['ancetre_commun']}")
    print("Chemin de traverse inter-domaines :")
    for step in multi_step_query.get('chemin_le_plus_court', [])[:10]:
        print(f"  {step}")
else:
    print("Aucun chemin ontologique direct trouvé entre ces deux domaines disparates.")
```

### 🪞 Introspection et Représentation Visuelle de Soi
GrandNexus cartographie ses nœuds cognitifs les plus denses pour se représenter lui-même.

```python
import matplotlib.pyplot as plt
import networkx as nx

def visualize_self_map(agent, max_nodes=40):
    print("🧠 === Méta-Cognition : Carte Réflexive de Soi ===")
    nodes = list(agent.graph.nodes)
    concept_strength = {n: len(agent.graph.neighborhood(n)) for n in nodes}

    # Sélectionner les nœuds les plus centraux (le noyau de sa pensée)
    top_nodes = sorted(concept_strength.keys(), key=lambda x: concept_strength[x], reverse=True)[:max_nodes]

    G = nx.Graph()
    for n in top_nodes:
        G.add_node(n, size=concept_strength[n])

    for rel, pairs in agent.graph.edges.items():
        for s, t in pairs:
            if s in top_nodes and t in top_nodes:
                G.add_edge(s, t, label=rel)

    plt.figure(figsize=(12, 10))
    pos = nx.spring_layout(G, k=0.8, iterations=50)

    sizes = [G.nodes[n]['size'] * 30 for n in G.nodes]
    nx.draw_networkx_nodes(G, pos, node_size=sizes, node_color='cyan', alpha=0.6, edgecolors='blue')
    nx.draw_networkx_labels(G, pos, font_size=9, font_family='sans-serif')
    nx.draw_networkx_edges(G, pos, alpha=0.3, edge_color='gray')

    plt.title("GrandNexus - Carte Cognitive Réflexive (Core Nodes)", fontsize=14)
    plt.axis('off')
    plt.show()

# Génération de la carte sur l'agent avancé
visualize_self_map(advanced_agent, max_nodes=40)
```

### 🌊 Immersion Profonde (1 Heure) & Apprentissage Auto-Dirigé
Le système assimile massivement pendant 50 minutes, puis analyse sa propre structure pour déduire ses « préférences » (les nœuds les plus denses de son graphe) et oriente ses lectures dessus pendant les 10 dernières minutes.

```python
import time
import random
import concurrent.futures
import wikipedia
import arxiv

def preferred_worker(agent, theme):
    """Worker spécialisé qui cherche spécifiquement le thème préféré."""
    try:
        if random.random() > 0.5:
            search_res = wikipedia.search(theme, results=1)
            if not search_res: return None
            page = wikipedia.page(search_res[0], auto_suggest=False)
            text = page.summary
            prefix = f"[Wiki-Focus: {theme[:15]}]"
        else:
            search = arxiv.Search(query=theme, max_results=1)
            paper = next(arxiv_client.results(search))
            text = paper.summary.replace('\n', ' ')
            prefix = f"[Arxiv-Focus: {theme[:15]}]"

        triplets = extract_triplets(text)
        if triplets:
            with graph_lock:
                for sujet, relation, objet in triplets:
                    agent.graph.add(sujet, objet, relation)
            return f"🌟 {prefix} Appris : {len(triplets)} faits."
        return None
    except Exception:
        return None

def run_deep_immersion(agent, total_minutes=60, self_directed_minutes=10, max_threads=6):
    phase1_minutes = total_minutes - self_directed_minutes
    print(f"🌊 Lancement de l'Immersion Profonde ({total_minutes} minutes)")

    # ==========================================
    # PHASE 1 : Apprentissage généraliste massif
    # ==========================================
    print(f"\n▶️ PHASE 1 : Apprentissage généraliste ({phase1_minutes} minutes)")
    end_phase1 = time.time() + (phase1_minutes * 60)

    with concurrent.futures.ThreadPoolExecutor(max_workers=max_threads) as executor:
        while time.time() < end_phase1:
            futures = []
            for _ in range(max_threads):
                source = random.choice(["wikipedia", "wikipedia", "arxiv"])
                futures.append(executor.submit(enhanced_worker, agent, source))

            for future in concurrent.futures.as_completed(futures):
                res = future.result()
                if res:
                    # Affichage allégé pour ne pas faire crasher le navigateur sur 1h
                    print(f"[P1] {res[:80]}...")
            time.sleep(1)

    # ==========================================
    # PHASE 2 : Apprentissage auto-dirigé
    # ==========================================
    print(f"\n▶️ PHASE 2 : Apprentissage auto-dirigé (Préférences du graphe) ({self_directed_minutes} minutes)")

    # Méta-cognition : trouver les thèmes préférés (topologie dense)
    nodes = list(agent.graph.nodes)
    concept_strength = {n: len(agent.graph.neighborhood(n)) for n in nodes}
    top_nodes = sorted(concept_strength.keys(), key=lambda x: concept_strength[x], reverse=True)[:10]

    # Nettoyage des noms WordNet (ex: 'car.n.01' -> 'car')
    preferred_themes = [str(n).split('.')[0].replace('_', ' ') for n in top_nodes]
    print(f"\n🎯 Thèmes 'préférés' déduits par densité topologique :")
    print(preferred_themes)
    print("Le système oriente maintenant ses capteurs sur ces sujets précis...\n")

    end_phase2 = time.time() + (self_directed_minutes * 60)
    with concurrent.futures.ThreadPoolExecutor(max_workers=max_threads) as executor:
        while time.time() < end_phase2:
            futures = []
            for _ in range(max_threads):
                theme = random.choice(preferred_themes)
                futures.append(executor.submit(preferred_worker, agent, theme))

            for future in concurrent.futures.as_completed(futures):
                res = future.result()
                if res:
                    print(f"[P2] {res}")
            time.sleep(2)

    print("\n✅ IMMERSION TERMINÉE. Le graphe a atteint un nouveau stade de maturité.")
    print(f"📊 Taille finale du graphe : {agent.graph.count()} relations.")

# Exécution de la longue session d'une heure
# Vous pouvez modifier les valeurs si vous souhaitez raccourcir le test
run_deep_immersion(advanced_agent, total_minutes=60, self_directed_minutes=10, max_threads=8)
```
