```python
from google.colab import files
import zipfile
import os

# 1. Uploader le fichier depuis votre ordinateur
print("Veuillez sélectionner le fichier ZIP de votre IA : ")
uploaded = files.upload()

# 2. Dézipper le fichier téléchargé
dossier_destination = "ia_environnement"

for filename in uploaded.keys():
    if filename.endswith(".zip"):
        print(f"\nDécompression de {filename} en cours...")
        with zipfile.ZipFile(filename, 'r') as zip_ref:
            zip_ref.extractall(dossier_destination)
        print(f"\nSuccès ! L'IA a été extraite dans le dossier : ./{dossier_destination}")
    else:
        print(f"\nLe fichier {filename} n'est pas un fichier ZIP.")
```

```python
import os

dossier_cible = "ia_environnement"
print(f"Arborescence des fichiers dans '{dossier_cible}' :\n")

for root, dirs, files in os.walk(dossier_cible):
    # Calcul du niveau d'indentation pour l'affichage en arbre
    level = root.replace(dossier_cible, '').count(os.sep)
    indent = ' ' * 4 * (level)
    print(f"{indent}📁 {os.path.basename(root)}/")

    subindent = ' ' * 4 * (level + 1)
    for f in files:
        print(f"{subindent}📄 {f}")
```

```python
from google.colab import files
import json

print("Veuillez sélectionner le fichier 'GrandNexus_banc_disjonction.ipynb' depuis votre ordinateur :")
uploaded_banc = files.upload()

for filename in uploaded_banc.keys():
    if filename.endswith('.ipynb'):
        print(f"\nAnalyse du notebook {filename}...")
        with open(filename, 'r', encoding='utf-8') as f:
            notebook_data = json.load(f)

        cells = notebook_data.get('cells', [])
        print(f"Le notebook contient {len(cells)} cellules.\n")

        print("Aperçu des cellules :")
        for i, cell in enumerate(cells):
            cell_type = cell.get('cell_type', 'inconnu')
            source = cell.get('source', [])
            # Joindre les lignes et limiter pour l'aperçu
            preview = "".join(source)[:80].replace('\n', ' ')
            print(f" - Cellule {i+1} [{cell_type}] : {preview}...")
    else:
        print(f"\nAttention: Le fichier {filename} n'a pas l'extension .ipynb.")
```
