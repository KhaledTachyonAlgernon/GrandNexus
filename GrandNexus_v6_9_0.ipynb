```python
import zipfile
import os

zip_path = '/content/GrandNexus_v6.9.0-rc_nexus-logic_complet.zip'
extract_dir = '/content/GrandNexus_extracted'

# Créer le dossier d'extraction s'il n'existe pas déjà
os.makedirs(extract_dir, exist_ok=True)

# Décompresser le fichier ZIP
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_dir)

print(f"Le fichier a été décompressé avec succès dans le dossier : {extract_dir}")

# Afficher le contenu du dossier décompressé
print("\nContenu du dossier :")
print(os.listdir(extract_dir))
```

```python
import json

notebook_path = '/content/GrandNexus_banc_FOLIO.ipynb'

print("--- CODE EXTRAIT DE GRANDNEXUS_BANC_FOLIO.IPYNB ---\n")
try:
    with open(notebook_path, 'r', encoding='utf-8') as f:
        nb_content = json.load(f)

    for i, cell in enumerate(nb_content.get('cells', [])):
        if cell['cell_type'] == 'code':
            print(f"# Cellule de code {i+1}")
            print("".join(cell.get('source', [])))
            print("\n# " + "-"*40 + " \n")
except Exception as e:
    print(f"Erreur lors de la lecture du fichier : {e}")
```
